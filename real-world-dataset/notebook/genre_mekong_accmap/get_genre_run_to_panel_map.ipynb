### Background

The GenRe Mekong data has all been submitted to ENA, with each sample being associated with 3 raw data products - one for each of the 3 amplicon sub-panels (GRC1, GRC2, SPEC). Due to way that the data was submitted, the information on which sub-panel a given ENA run object corresponds to is only available in one place: the "submitted_ftp" property. E.g.:

```python

import sys
from pathlib import Path
sys.path.append(str(Path("../../../lib").resolve()))

from ENA_data_helper import create_ena_data_frame_from_bioproject

df = create_ena_data_frame_from_bioproject("PRJEB85801")
df
```

### The Problem

ENA announced at the end of 2025 that they would be withdrawing the ability to download submitted sequence files (fastqs and bams). This process will begin on 31st March 2026, at which point the submitted_ftp property will return a NULL value. After this date, this crucial piece of metadata will be lost! We therefore will record this critical metadata here for posterity.  

```python

df["panel"] = (
    df["submitted_ftp"]
    .str.extract(r"(GRC1|GRC2|SPEC)", expand=False)
    .fillna("Undefined")
)

df["panel"].value_counts()
```

```python
df = df[['sample_title','run_accession','panel']]
df.sort_values(by=["sample_title","run_accession"]).to_csv('genre_ena_runacc_to_panel_map.csv', index=False)
```
